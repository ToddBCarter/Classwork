### Part 3: Functions and Modules

We've been using function for quite a while and now we discuss it officially. Functions are a fundamental programming construct that serves two primary purposes: promoting code reusability and facilitating procedural decomposition.

1. Code Reuse and Redundancy Reduction: Functions provide a mechanism to encapsulate a block of logic that may be used repeatedly. By defining a task within a function, it can be invoked multiple times throughout a program, which maximizes code reuse and minimizes redundancy.

2. Procedural Decomposition: Functions allow for the breakdown of complex systems into smaller, more manageable, and well-defined modules. This modular approach improves code organization, readability, and maintainability. For example, the implementation of a machine learning program can be decomposed into a series of distinct functions, each responsible for a specific subtask.

#### Positional Arguments

As the preceding examples illustrate, the sequence in which arguments are passed to a function is critically important. While combining positional and keyword arguments is permissible, it can sometimes be a source of confusion.

Positional arguments are the most conventional method for passing values. When using this approach, the supplied arguments are mapped to the function's defined parameters sequentially from left to right. Therefore, the order in which they are provided is essential for correct assignment.

#### Keyword Arguments

With keyword arguments, you explicitly assign a value to a parameter by name (`parameter=value`). Because you name them, the order does not matter.

#### Combining Positional and Keyword Arguments

You can use both types in a single function call, but you must follow one critical rule:

- All positional arguments must come before any keyword arguments.

#### Parameter vs. Argument

The terms parameter and argument are often used interchangeably, but they have distinct meanings in programming.

The simplest way to think about it is:

- A Parameter (formal parameter) is the variable listed inside the parentheses in a function's definition. It's a placeholder.
- An Argument (actual parameter) is the actual value that is sent to the function when it is called. It's the real deal.

#### Functions with `*args' and `**kargs'

In Python, *args and **kwargs are special syntax used in function definitions to pass a variable number of arguments.

- `*args` collects extra positional arguments into a tuple.
- `**kwargs` collects extra keyword arguments into a dictionary.

When you use `*args` as a parameter, it allows your function to accept any number of positional arguments. The `*` symbol "unpacks" the arguments, and the name args (short for arguments) is just a convention.

Similarly, `**kwargs` allows your function to accept any number of keyword arguments (e.g., name="Alice"). The ** symbol collects these into a dictionary, where the keys are the parameter names and the values are the arguments. The name kwargs (short for keyword arguments) is the convention.

#### Docstrings
A docstring is a string literal that occurs as the first statement in a module, function, class, or method definition. It is used to document what the code does. Good documentation is key to writing understandable and maintainable code. You can access a function's docstring using the built-in `help()` function or the `__doc__` attribute.

·#### Type Hinting

Type hints are a feature that allows you to indicate the expected data types of variables, function parameters, and return values. They do not affect how the code runs, but they provide valuable information for developers and tools like code editors and static analyzers.

You add them using a colon `:` after the parameter name and `->` for the return value.

### 3.2 Namespace

In Python, a namespace is a system that keeps names (like variables and functions) unique and organized to avoid conflicts. The concepts of __main__, built-in functions, and function namespaces all relate to scope, which is the set of rules Python uses to look up a name. Python resolves names by checking different namespaces in a specific order, known as the LEGB rule:

- Local
- Enclosing
- Global
- Built-in

#### Local (Function) Namespace

This is the namespace inside a function. It's created when the function is called and forgotten when the function ends. It contains all the function's parameters and any variables defined within it. Now let's see an example in which you can not access a local (function) variable outside its scope.

#### Enclosing Namespace

This special scope exists for nested functions. The inner function `can` access names from the outer (enclosing) function's namespace.

#### Global (__main__) Namespace

This is the main namespace of your script. Any name you define at the top level of a `.py` file (not inside a function or class) is in the global namespace. When you run a script directly, Python gives this global namespace the special name `__main__`.

#### Built-in Namespace

This is the outermost namespace that contains all of Python's built-in functions and exceptions, like `print()`, `len()`, `str()`, and `ValueError`. These names are always available in any scope.

#### `global` Keyword

To modify a variable in the global scope from within a function, you must explicitly declare your intent using the global keyword at the beginning of the function.

#### `nolocal` Keyword

The `nonlocal` keyword is used in nested functions to indicate that a variable refers to a variable in the nearest enclosing scope, which is not global.

In simple terms, it allows an inner function to modify a variable that belongs to the outer function. Without nonlocal, assigning a value to that variable would create a new, separate variable in the inner function's local scope.

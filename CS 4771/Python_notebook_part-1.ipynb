#### Python's Basic Data Types
To work correctly, Python needs to know what kind of data it's dealing with. After all, you can do math with a temperature, but you can't do math with a name! This **"kind of data"** is called a data type. Python has several basic types built right in:

- Integer (int): These are whole numbers, both positive and negative, without any decimal points. Think of things you can count, like 100 points on a test or 3 apples.

- Float (float): These are numbers that do have a decimal point, used for more precise measurements. Examples include a weight of 150.5 lbs or a price of $19.99.

- String (str): This is just plain text. To let Python know you're using a string, you wrap the text in single (**' '**) or double (**" "**) quotes. Examples include "Hello, world!" or 'Python is fun'.

#### Strings and operations
Python strings (type str) are enclosed in single quotes ('...'), double quotes ("..."), or triple quotes ("""...""").

We'll see that various types of parentheses are used extensively in Python, but why do quotation marks seem a bit redundant?

The first scenario is when you use single and double quotes interchangeably to avoid escaping the quotes themselves.

The second case is using triple quotes to create multi-line strings while preserving their formatting.

#### String Operations
Python supports various string operations. Let's take a look at them one by one.

First, concatenation and splitting are probably the most commonly used. 

- The simplest way to concatenate strings is by using the + operator.

- .join(iterable): This method connects all the strings in an iterable object (such as a list) using the specified string as a separator. It is the most recommended approach for concatenating a large number of strings.

- .split(separator): This method divides a string into a list of substrings based on a specified separator.

- A special application is using the * operator to repeat a string.

Similarly, finding and replacing are fundamental string operations. 

- in operator: This operator checks if a substring is present within a string.

- .find(substring): This method returns the index of the first occurrence of a substring. It returns -1 if the substring is not found.

- .replace(old, new): This method replaces every occurrence of the substring old with the substring new.

Building on the inherent characteristics of strings, Python provides built-in methods for case conversion and formatting.

- .lower() and .upper(): These methods convert a string to all lowercase or all uppercase.

- .strip(): This method removes leading and trailing whitespace characters from a string. .lstrip() and .rstrip() remove whitespace only from the left and right sides, respectively.

- .capitalize(): This method converts the first character of a string to uppercase.

- .title(): This method capitalizes the first letter of each word in a string.

Different prefixes can be used to create strings with various effects. Here, we'll introduce the most common ones: raw strings, binary strings, and f-strings.

- When you prefix a string with r, Python treats the backslash (\) as a literal character rather than an escape character.

- In Python, the prefix b is used to denote a byte string, which is an immutable sequence composed of binary bytes ranging from 0 to 255. This data type is primarily used for handling raw binary data, such as in file I/O and network communication.

- The f-string prefix in Python is a modern and concise method for string formatting. It allows you to embed expressions directly within a string literal.

Since strings are a type of sequence, they can be indexed and sliced. Although indexing and slicing are more extensively discussed when introducing lists, strings are an excellent example of these operations.

Each character within a string has a unique index, which is its position numbered from left to right. Python offers both forward (or positive) and backward (or negative) indexing.

- Forward indexing (or positive indexing) is a method used to access elements in a sequence, such as a string or a list, by their position starting from the beginning. The first element is at index 0, the second is at index 1, and so on.

- Backward (or negative) indexing is a method of accessing elements from the end of a sequence. The last element is at index -1, the second to last is at index -2, and so on.

Slicing is a powerful and flexible operation in Python that allows you to extract a subsequence from sequence types like strings, lists, and tuples. The core of slicing lies in its concise syntax.

In Python, the slicing operation allows for extracting a portion of a sequence by specifying a range of indices.  The syntax for this operation is：

$sequence[start:stop:step]$ 

The start parameter defines the inclusive starting index, while the stop parameter specifies the exclusive ending index. The step parameter determines the interval at which elements are selected. All three parameters are optional: start defaults to 0, stop defaults to the end of the sequence, and step defaults to 1.

There are many useful slicing techniques, which will be demonstrated in the following lectures and implementations.

#### List

In Python, lists are the most flexible **ordered** built-in collection object type.
```python
    L = [1, 2, 3, 5] # a list example
```

Because we already have some experience with strings, it's easier to understand lists. A list is also a sequence type, so it supports indexing and slicing. Unlike strings, lists in Python are mutable and can contain different data types.

First, let's create a list just like we did in the example above.

Next, let's go over the common operations for the list data type.

- We can use the len() method to get the length of a list, and it also works for strings.



- We can concatenate lists by using either the + operator or the .extend() method.

- Lists also support the * operator to perform a repeat operation.

- The in membership method is also applicable for lists.

- Because lists are mutable, we can use the append() method to add an element to the end of a list.

- list.insert(index, element) is a highly useful method in Python for inserting an element at a specified position within a list.

- For mutable objects like lists, deletion operations are indeed essential. Python offers several distinct methods for removing elements from a list. del is a versatile statement that can be used to delete variables, as well as specific elements or slices from a list.

- The .remove() method is used to delete elements by their value. It removes the first element that matches the specified value.

- The .pop() method is used to remove an element by its index, and it also returns the element that was removed. This makes it ideal for situations where you need to delete and retrieve the value of the removed element simultaneously, such as when implementing a stack or a queue.

- Alternatively, you can clear delete elements from a list using the slice assignment a[0:1] = []

- List comprehensions are often considered one of the best features of the list data type. While a standard loop can achieve the same result, list comprehensions are generally more efficient and lead to more readable code.

#### Tuple

A tuple is a collection of objects which is ordered and immutable, and it is commonly written as a series of items in parentheses (). Simply put, a tuple is an immutable list. Most operations that work for list also work for tuple type.

- Similar to lists, tuples can contain elements of different data types.

- We can use the len() method to get the length of a tuple.

- Tuples also support indexing and slicing operations.

- Similar to lists, tuples also support concatenation.

A frequently asked question is what the purpose of a tuple is. There are several reasons for this. Thanks to their immutability, tuples can be used as keys in a dictionary, whereas lists cannot. Tuples are also commonly used to return multiple values from a function.

- Tuples can be created without using parentheses. This behavior is not a specific design feature but is a direct consequence of Python's logic desiging tuple.

- Leveraging this feature, we can also easily achieve multiple assignment.

#### Dictionary

Flexible **unordered** built-in collection object type. Items are sorted and fetched by **key** instead of by positional offset.
```python
    D = {'name': 'Tim', 'age': 22}
    print(D)

    print(D['name']) # (1) accessed by key
    D['age'] = 23 # (4) mutable
    print(D)
```

A Python dictionary is a mutable, variable-length collection of unsorted objects, implemented as a hash table, where each item is accessed by a unique key.

Traditionally, dictionaries were considered unordered data structures. However, since Python 3.7, the language specification now guarantees that dictionaries preserve the insertion order of their items. While this behavior was an implementation detail in Python 3.6, it is now a core feature of the language.

- First, let's create a dictionary.

- We can still use the built-in len() function to get the number of key-value pairs in a dictionary.

- Because dictionary keys are not as straightforward as an index, Python provides the in operator to check if a key exists within a dictionary.

- Python dictionaries provide three useful methods for retrieving all of their keys, values, or key-value pairs.

- We can remove elements from a dictionary using either the del statement or the .pop() method.

Dictionaries also support comprehensions, a feature built on top of Python's iterators. The specific implementation details will be discussed in more detail later.

#### Set

A ***set*** is a collection of **unique objects** which is **unordered and immutable**, and are constructed by using the set() function.

Sets support operations corresponding to mathematical set theory (intersection, union, etc.). By definition, an item appears only once in a set, no matter how many times it is added. 

Because sets are collections of objects, they share some behavior with lists and dictionaries. For example, sets are iterable, can grow and shrink on demand, and may contain a variety of object types. 

However, since sets are unordered and do not map keys to values, they are neither a sequence (list) or mapping type (dict).

Sets have a variety of applications, especially in numeric and database-focused work.

- First, let's create a set. We'll choose to create it from a list.

- Given the nature of sets, duplicate elements are automatically removed when you create a set using a factory function. In fact, this method is quite common.

- We can use the - operator to find items that are in set1 but not in set2.

- We can use the | operator to union two sets.

- We can use & operator to find the intersection of two sets.

- We can use add method to add a item to a set.

- We can use remove method to remove a item to a set.

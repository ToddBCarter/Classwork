### Part 2: Flow Control

Although `built-in` data types offer powerful capabilities for solving problems, they are often insufficient for applications that require more dynamic behavior. For instance, when code is intended for distribution, it must be designed to accept a range of inputs and adapt its execution accordingly. A tax calculator provides a classic example, as its computations vary based on the input income level. Therefore, to implement such conditional logic, it is necessary to introduce more sophisticated statements for controlling the program's flow of execution.

Flow control (or control flow) refers to the order in which statements and instructions are executed. Think of it like a recipe: instead of following every step in order, you might have instructions like "if using fresh tomatoes, peel them first" or "whisk for 5 minutes."

In Python, the primary flow control tools are:

- Conditional Statements: Execute code only if a certain condition is true.

    `if`, `elif`, `else`

- Loops: Repeat a block of code multiple times.

    `for` (iterating over a sequence)

    `while` (looping as long as a condition is true)

- Loop Modifiers: Change a loop's behavior from within.

    `break` (exits the loop entirely)

    `continue` (skips the rest of the current iteration and moves to the next)

    `pass` (a placeholder that does nothing)

- Exception Handling: Manage errors and other exceptional events.

    `try`, `except`, `finally`, `else`

- Function and Generator Calls: Transfer control to a function and (usually) back.

    `def`, `return`, `yield`

- Structural Pattern Matching: (Python 3.10+) A sophisticated conditional structure similar to a switch-case statement.

    `match`, `case`

#### 2,1 The Boolean Data Type

The Boolean data type (`bool`), which was not addressed in the preceding overview of Python's data types, is used to represent logical truth values. This type is limited to two constant values: `True` and `False`.

Aside from directly using the `True` and `False` constants, boolean values are commonly generated through the following methods:

- comparison operations
- logical operations
- built-in functions

Comparison operations evaluate the relationship between two values and return a boolean result: either `True` or `False`. They're fundamental for decision-making and controlling the flow of a program. Common Comparison Operators include:

- `==` : Equal to (e.g., `x == y`)

- `!=` : Not equal to (e.g., `x != y`)

- `\>` : Greater than (e.g., `x > y`)

- `<` : Less than (e.g., `x < y`)

- `\>=` : Greater than or equal to (e.g., `x >= y`)

- `<=` : Less than or equal to (e.g., `x <= y`)


The identify operators compare the memory location of two objects.

- `is` : Returns `True` if both variables point to the same object.
- `is not` : Returns `True` if both variables point to different objects.

Logical operators are used to combine conditional statements, typically with Boolean values.

- `and` : Returns `True` if both statements are true.
- `or` : Returns `True` if at least one of the statements is true.
- `not` : Reverses the result, returning `False` if the result is true, and vice-versa.

Membership operator test if an object is present in a sequence.

- `in`: Returns `True` if a value is found in the sequence.
- `not in`: Returns `True` if a value is not found in the sequence.

The built-in functions `bool()` and `isinstance()` both return Boolean values (`True` or `False`).

#### 2.2 `if` Statements

The Python `if` statement is a conditional statement that runs a `block` of code only when a specific condition is evaluated as `True`. Compared to C++, Python uses a fundamentally different syntax to structure if statements and define their associated code blocks..

C++ example of the 'if' statement
```c++
if(a > b){
    x = 1;
    y = 2;
}
```

Python example of the 'if' statement
```python
if a > b:
    x = 1
    y = 2
```

Things to notice:
- How the code blocks are defined
- How to end the hider line

A simple `if` statement executes a block of code only when a specific condition is `True`.

```
if condition:
    statement 1
    statement 2
```

You'll often need to run one block of code if a condition is `true` and a different block if it's `false`.

```
if condition:
    statement 1
    statement 2
else:
    statement 3
    statement 4
```

For situations where you need to check several conditions in a row, Python provides the `elif` statement.

```
if condition 1:
    statement 1
elif condition 2:
    statement 2
elif condition 3:
    statement 3
else:
    statement 4
```

Python offers a more compact way to write a simple `if-else` statement on a single line using a feature called a conditional expression or ternary operator.

While it can make your code shorter, the ternary operator is best used for simple, clear-cut assignments. If the logic is complex or the expressions are long, a standard `if-else` block is often more readable and a better choice.

#### 2.3 `while` Loops

A Python `while` loop repeatedly executes a block of code as long as a given condition remains `True`. It's useful when you don't know in advance how many times the loop needs to run. The structure of a while loop is straightforward. The loop's logic follows a simple cycle: check the condition, and if it's `True`, run the code block, then repeat.
- The condition is tested at the beginning of each iteration.
- If the condition is `True`, the indented code block is executed.
- After the block runs, the condition is tested again.
- This process repeats until the condition becomes `False`, at which point the loop terminates and the program continues with the code that follows.

The general syntax of a basic while loop:
```python
while condition:
    statement 1
    statement 2
```

A common pitfall is the infinite loop, which occurs if the loop's condition never becomes `False`. This will cause your program to hang. You can stop it manually by pressing Ctrl+C in the terminal.

You can control the flow of a while loop from within the code block using two key statements:

- `break`: Immediately terminates the loop, regardless of the condition.
- `continue`: Skips the rest of the current iteration and jumps back to the top to re-evaluate the loop's condition.

#### 2.4 `for` Loops

A Python `for` loop is used for iterating over a sequence (like a list, tuple, string, or range of numbers). It's the go-to loop when you want to execute a block of code for each item in a collection.


Syntax of the *for* Loop

- The Python for loop starts with the keyword "for" followed by an arbitrary variable name, which will hold the values of the following data sequence. The *else* block will be executed only if the loop hasn't been "broken" by a break statement.

```python
for item in iterable:
    statement 1
    statement 2
```

Python's iterator mechanism is the protocol that allows for loops and other tools to step through sequences one item at a time. It's built on two distinct concepts: iterables and iterators.

The `range()` function is the standard and most efficient way to create number-based for loops in Python. The `range()` function can be used in three main ways

- `range(stop)`
- `range(start, stop)`
- `range(start, stop, step)`

#### 2.5 Match

The `match` statement, introduced in Python 3.10, provides a powerful form of structural pattern matching that is similar to a `switch` statement in many other languages. The structure consists of a `match` keyword followed by the subject expression. Below it are one or more case blocks, each with a pattern to match against.

```python
match subject:
    case pattern_1:
        # Code to execute if subject matches pattern_1
    case pattern_2:
        # Code to execute if subject matches pattern_2
    case _:
        # A wildcard pattern that matches anything (like a default case)
```

The real power of `match` comes from its ability to match the structure of an object, not just its value. For example, you can combine multiple literal values into a single case using the `|` (OR) operator.

You can match against sequences like tuples or lists and "capture" parts of them into variables.

You can add an if condition, known as a guard, to a case for an even more specific match.

#### 2.6 Exception Handling

Python handles errors using exception handling, a mechanism that allows you to manage runtime errors gracefully instead of letting them crash your program. This is done primarily with the `try...except` block.
- `try`: You place the code that might cause an error inside the try block.
- `except`: If an error (an exception) occurs in the try block, Python immediately stops executing that block and jumps to the except block. The code in the except block is then executed to handle the error.

Without exception handling, code will crash when it encounters an error.

You can handle all exceptions by using a bare `except:` block, but this is strongly discouraged as it can hide serious problems and make your code difficult to debug.

The correct way to use `except Exception as e:` to catch most general, non-system-exiting exceptions is to explicitly catch the base `Exception` class. This is the parent class for nearly all built-in, non-system-exiting exceptions.

It's best practice to specify the type of exception you expect to catch. This prevents your code from accidentally catching unrelated errors. You can handle multiple specific errors by adding more except blocks or by grouping them.

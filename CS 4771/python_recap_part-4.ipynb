### Part 4: Object Orient Programming (OOP)

Object-Oriented Programming (OOP) is a way of structuring software around "objects" rather than functions and logic. An object bundles together related data (called attributes) and the functions that operate on that data (called methods).

This approach helps you create modular, reusable, and organized code by modeling real-world things. For example, you could model a car as an object with attributes like color and model and methods like `drive()` and `brake()`.

### 4.1 Classes and Objects

Think of a class as a blueprint or a template. It defines the properties (called attributes) and behaviors (called methods) that all objects of that type will have.

An object (or instance) is a specific creation based on that blueprint. For example, Dog could be a class, while your specific dog, Fido, is an object of that class.

Why use them? Classes are the foundation of Object-Oriented Programming (OOP), which helps you organize complex code into reusable and logical structures.

#### 4.2 Class Attributes vs. Instance Attributes

- Instance Attributes: Belong to a specific instance of a class (like self.name). Each object has its own copy. They are defined inside __init__.
- Class Attributes: Shared by all instances of the class. They are defined directly under the class declaration.

#### 4.2 Inheritance
Inheritance allows you to create a new class (the child class) that inherits the attributes and methods of an existing class (the parent class). This promotes code reuse and establishes a clear relationship between classes.

The `super()` function is used to call methods from the parent class, which is useful for extending, not just replacing, parent behavior.

- If a child class doesn't have its own `__init__` method, it automatically inherits and uses the `__init__` from its parent.
- If you define an `__init__` method in the child class, you are overriding (replacing) the parent's `__init__`. It will not be called automatically. You must explicitly call it using `super()` if you want its logic to run.

In short: If you override `__init__`, you take full responsibility for initialization, which almost always includes calling super() to let the parent do its part first.

#### 4.3 Special Methods

You may wonder why the initialization function must be named as `__init__`. Python classes have special methods, often called "dunder" (double underscore) methods, that start and end with double underscores. They allow you to define how your objects behave with built-in Python operations. Here are some examples of special (magic) methods:

- `__str__(self)` for `print()` and `str`
- `__len__(self)` for `len()`
- `__eq__(self, other)` for `==`

#### 4.4 Encapsulation

Encapsulation is the bundling of data (attributes) and methods that operate on that data into a single unit, the object. It hides the object's internal state and complexity from the outside, only exposing what's necessary.

#### Attribute

- Public: Accessible from anywhere (e.g., `self.name`).
- Protected: By convention, attributes prefixed with a single underscore `_` are treated as non-public. It's a hint to other developers not to access it directly from outside the class (e.g., `self._balance`).
- Private: Attributes prefixed with a double underscore `__` are name-mangled by Python, making them harder to access from outside (e.g., `self.__account_number`).

Python does not have a concept of truly private class attributes. Instead, it employs a mechanism known as "name mangling" to make it more difficult to inadvertently access such attributes from outside the class. The effect of this name alteration can be observed by using the built-in `dir()` function on an object instance.

#### Instance Methods

Instance methods are the most common type and are bound to a specific object (instance). They receive the instance as their first argument, conventionally named self.
- Can Access Instance Attributes? Yes. This is their primary purpose. They use self to read and modify the unique attributes of that specific instance (e.g., self.name).
- Can Access Class Attributes? Yes. They can access attributes shared by all instances of the class using self.__class__.attribute or ClassName.attribute.

#### Class Methods

Class methods are bound to the class itself, not a specific instance. They are marked with the `@classmethod` decorator and receive the class as their first argument, conventionally named cls. They are often used as factory methods to create instances in specific ways.

- Can Access Instance Attributes? No. They do not receive a self argument, so they have no way of knowing about any specific instance's attributes.
- Can Access Class Attributes? Yes. This is their main advantage. They use cls to read or modify attributes that are shared across all instances of the class.

#### External Access of Attributes

External code can access the attributes of a class through an instance of that class, but the access is governed by the attribute's naming convention (public, protected, or private).

-Public Attributes: These are fully accessible. External code can freely read and modify them.
-Protected Attributes `_`: These are technically accessible, but it's a strong convention that they should not be touched from outside the class. Doing so is considered bad practice.
-Private Attributes `__`: These are not directly accessible from outside. Python's name mangling renames the attribute, causing a direct access attempt to fail with an AttributeError.

#### External Access of Methods

External code can call the methods of a class through an instance, with access being determined by the method's naming convention (public, protected, or private).

- Public Methods: These are part of the class's public API and are meant to be called from anywhere. This is the standard way to interact with an object.
- Protected Methods `_`: While technically callable from outside, the single underscore is a strong convention that signals the method is for internal use by the class or its subclasses. Calling a protected method from external code is considered bad practice.
- Private Methods `__`: These are not directly callable from the outside. Python's name mangling prevents external access, which will result in an AttributeError.

#### Polymorphism

This principle allows objects of different classes to be treated as objects of a common superclass. It means "many forms," and it lets you call the same method on different objects, and each object will respond in its own way.

#### Polymorphism through Inheritance

The classic example of polymorphism involves inheritance, where multiple child classes inherit from a parent class and provide their own specific implementations of a shared method.

#### Polymorphism and Duck Typing

Python's dynamic nature allows for a more flexible form of polymorphism called duck typing. The name comes from the saying: "If it walks like a duck and it quacks like a duck, then it must be a duck."

This means that in Python, the type or class of an object is less important than the methods it defines. If an object has the required method, it can be used in the operation, regardless of its inheritance.

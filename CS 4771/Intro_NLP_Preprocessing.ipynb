# NLP Introduction: Preprocessing

# 1. What is Natural Language Processing (NLP)?

NLP is a field of artificial intelligence (AI) that focuses on the interaction between computers and humans using natural language. The ultimate goal of NLP is to enable computers to read, understand, interpret, and generate human language in a way that is both valuable and meaningful.

It is interesting that NLP and ML have not always overlapped significantly. Previously, teaching NLP required discussing ML, but teaching ML did not require discussing NLP. This is largely because the development of large models in recent years has given NLP an increasingly important influence within ML, which in turn promotes the development of ML and related fields like Computer Vision.


NLP powers many applications we use daily, such as:
* Spam filters in your email
* Virtual assistants (Siri, Alexa, Google Assistant)
* Language translation (Google Translate)
* Autocorrect and predictive text
* Chatbots and customer service automation



## 1.1 Challenges

Human language is incredibly complex and nuanced, not built on a rigid set of rules like a programming language. This creates significant challenges:

* **Ambiguity:** The same word or phrase can have multiple meanings.
    * **Lexical Ambiguity:** "I'm going to the *bank*." (financial institution or river bank?)
    * **Syntactic Ambiguity:** "I saw the man *with the telescope*." (Who has the telescope? Me or the man?)
* **Context:** The meaning of a word is highly dependent on the surrounding text and the real-world situation. "That's sick!" can mean "That's disgusting" or "That's amazing," depending on the context.
* **Irony and Sarcasm:** The literal meaning of the words is the opposite of the intended meaning. Computers struggle to detect this.
* **Creativity and Evolution:** Language constantly evolves. New words are created (e.g., "selfie"), and spelling and grammar are informal (e.g., "lol", "brb").

## 1.2 Common NLP Tasks

NLP is used to solve a wide variety of problems. Here are some of the most common tasks:

* **Text Classification:** Assigning a category or label to a piece of text (e.g., sentiment analysis: positive/negative/neutral; topic classification: sports/tech/politics).
* **Named Entity Recognition (NER):** Identifying and categorizing key entities in text (e.g., "Google" -> ORGANIZATION, "London" -> LOCATION, "Dr. Smith" -> PERSON).
* **Machine Translation:** Automatically translating text from one language to another.
* **Question Answering (Q&A):** Providing a specific answer to a question based on a given context document.
* **Text Summarization:** Generating a concise and coherent summary of a longer document.
* **Text Generation:** Creating new, human-like text (e.g., chatbots, story writing with models like GPT).

## 1.3 The NLP Pipeline

To solve these tasks, we almost always follow a multi-step process known as the **NLP Pipeline**. While the exact steps can vary, a general pipeline looks like this:

1.  **Data Acquisition:** Gathering the raw text data (e.g., from files, web scraping, APIs).
2.  **Text Preprocessing:** Cleaning and structuring the raw text into a format suitable for machines.
3.  **Feature Extraction:** Converting the processed text into numerical representations (vectors) that models can understand (e.g., Bag-of-Words, TF-IDF, Word Embeddings).
4.  **Model Building:** Training a machine learning model on the numerical features to perform a specific task (e.g., classification).
5.  **Evaluation:** Assessing the model's performance on unseen data.

# 2. Text Preprocessing: The Essential First Step

Raw text is messy. It contains punctuation, capitalization, common but non-informative words, and different forms of the same word (e.g., "run", "running"). Preprocessing aims to clean and normalize this text to reduce its complexity and variability, making it easier for a model to find patterns.

We will cover the following key preprocessing steps:
1.  **Tokenization**
2.  **Normalization** (Lowercasing, Punctuation Removal)
3.  **Stop Word Removal**
4.  **Stemming & Lemmatization**

We'll use **NLTK (Natural Language Toolkit)**, a library that's great for granular, step-by-step control, and **spaCy**, a modern, high-performance library designed for building production-ready NLP pipelines.

Library Setup

```python
# Import the libraries
import nltk
import spacy
import string

# Load the spaCy model
nlp = spacy.load("en_core_web_sm")

# Download necessary NLTK data modules
# 'punkt' is for tokenization
# 'stopwords' is for the list of stop words
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')

print("Libraries and models loaded successfully!")
```

Our Sample Text

Let's define a raw text string that we can use to test all our preprocessing steps. It includes capitalization, punctuation, and multiple forms of the same word.

```python
raw_text = "Natural Language Processing (NLP) is an amazing and challenging field. It's been growing rapidly. We're studying running, runs, and ran."
```

## 2.1 Tokenization

**Tokenization** is the process of splitting text into smaller units called **tokens**. These can be words, characters, or subwords.


### 2.1.1 Word Tokenization
This is the most common form (not really), splitting text by words and punctuation.

```python
from pprint import pprint
from nltk.tokenize import word_tokenize

word_tokens = word_tokenize(raw_text)
pprint(word_tokens)
```

Notice how NLTK is smart enough to split "field." into `"field"` and `"."`, and "(NLP)" into `"("`, `"NLP"`, and `")"`.

### 2.1.2 Tokenization with spaCy

spaCy performs many steps at once. When you call `nlp(text)`, it automatically tokenizes the text and creates a `Doc` object. Each token is an object with rich linguistic annotations.

```python
doc = nlp(raw_text)

# We can get the text of each token
spacy_tokens = [token.text for token in doc]
pprint(spacy_tokens)
```

Note: spaCy's tokenizer is more robust and handles cases like "It's" (`"It"`, `"'s"`) and "We're" (`"We"`, `"'re"`) differently, often grouping them as single tokens with attached morphology, which is a more advanced approach.

### 2.1.3 Subword Tokenization (BPE, WordPiece)

This is a more advanced technique used by modern transformers (like BERT and GPT). It splits words into common sub-units. For example, "running" might become `"run"` and `"##ing"`.

This is useful because it can handle **Out-of-Vocabulary (OOV)** words. If the model has never seen "jetpacking," it can still understand it if it has seen `"jet"`, `"pack"`, and `"##ing"`.

We won't implement this from scratch, but it's crucial to know it exists.

```python
from transformers import GPT2Tokenizer

tokenizer = GPT2Tokenizer.from_pretrained('gpt2')

tokens = tokenizer.tokenize(raw_text)

pprint(tokens)
```

## 2.2 Normalization: Lowercasing and Removing Punctuation

**Normalization** is the process of converting all text to a consistent format. We want the model to understand that "Field" and "field" are the same word.

We'll work with our list of NLTK tokens (`word_tokens`).

Lowercasing

```python
lower_tokens = [token.lower() for token in word_tokens]
print("Lowercased:", lower_tokens)
```

Removing Punctuation. We can check if a token is alphanumeric (a-z, 0-9)

```python
normalized_tokens = [token for token in lower_tokens if token.isalnum()]
print("Normalized (Lower + No Punct):\n", normalized_tokens)
```

This list is much cleaner! We've removed all punctuation marks (`(`, `)`, `.`) and standardized the case.

## 2.3 Stop Word Removal

**Stop words** are extremely common words that add little semantic meaning to a sentence (e.g., "is", "an", "the", "it's"). We often remove them to help the model focus on the more important, topic-specific words.

NLTK provides a standard list of stop words.

```python
from IPython.display import Image
Image(url="https://media.cheggcdn.com/media/088/0881b012-dfb3-4716-ac8d-79513d377daf/phpFSk9Fx")
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
print("Sample stop words:", list(stop_words)[:10])

filtered_tokens = [token for token in normalized_tokens if token not in stop_words]
print("\nTokens after stop word removal:\n", filtered_tokens)
```

Look at that! The list is now much shorter and contains only the words that carry the most meaning.

## 2.4 Stemming vs. Lemmatization

This is the final step. Our list still contains `"running"`, `"runs"`, and `"ran"`, as well as `"studying"`. These are all related to the root words "run" and "study".

The goal is to reduce these words to their **base form** (or *lemma*). We have two main approaches:



### 2.4.1 Stemming (with NLTK)

**Stemming** is a crude, rule-based process. It chops off the ends of words to get to a common "stem". It's fast, but often the resulting "stem" is not a valid word.

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

stemmed_tokens = [stemmer.stem(token) for token in filtered_tokens]
print("Stemmed Tokens:\n", stemmed_tokens)

# Let's see how it handles our test words
print("\n--- Stemming Examples ---")
print(f"'studying' -> {stemmer.stem('studying')}")
print(f"'amazing'  -> {stemmer.stem('amazing')}")
print(f"'running'  -> {stemmer.stem('running')}")
print(f"'runs'     -> {stemmer.stem('runs')}")
print(f"'ran'      -> {stemmer.stem('ran')}")
```

Notice the results:
* `"studying"` -> `"studi"` (correctly groups, but "studi" isn't a word)
* `"amazing"` -> `"amaz"` (also not a word)
* `"running"` and `"runs"` -> `"run"` (Good!)
* `"ran"` -> `"ran"` (**Failure!**) Stemming is too simple to know that "ran" is the past tense of "run".


### 2.4.2 Lemmatization (with spaCy)

**Lemmatization** is a smarter, dictionary-based process. It uses morphological analysis and vocabulary context to find the true base form of a word, known as its **lemma**. It's slower than stemming but much more accurate.

spaCy's pipeline makes this easy. The `token.lemma_` attribute gives us the lemma for each token.

```python
doc = nlp(raw_text)

lemma_tokens = [token.lemma_ for token in doc if not token.is_stop and not token.is_punct and not token.is_space]

print("Lemmatized Tokens:\n", lemma_tokens)

print("\n--- Lemmatization Examples ---")
test_doc = nlp("studying amazing running runs ran")
for token in test_doc:
    print(f"'{token.text}' -> {token.lemma_}")
```

Look at the difference!

* `"studying"` -> `"study"` (A real word!)
* `"amazing"` -> `"amazing"` (Correctly identifies it as its own lemma)
* `"running"` and `"runs"` -> `"run"` (Good!)
* `"ran"` -> `"run"` (**Success!**) Lemmatization understands the linguistic connection between "ran" and "run".

**Conclusion:** Use **Lemmatization (spaCy)** if you need accuracy. Use **Stemming (NLTK)** if you need extreme speed and don't mind the loss of precision.

# 3. Full Preprocessing Pipelines

Let's combine what we've learned into two functions. Each function will take raw text and return a final list of processed tokens.

### NLTK Pipeline (Stemming)

```python
def nltk_pipeline(text):
    # 1. Tokenize
    tokens = word_tokenize(text)

    # 2. Normalize (lower, punc)
    lower_tokens = [token.lower() for token in tokens]
    normalized_tokens = [token for token in lower_tokens if token.isalnum()]

    # 3. Stop Word Removal
    stop_words = set(stopwords.words('english'))
    filtered_tokens = [token for token in normalized_tokens if token not in stop_words]

    # 4. Stemming
    stemmer = PorterStemmer()
    stemmed_tokens = [stemmer.stem(token) for token in filtered_tokens]

    return stemmed_tokens

print("NLTK Pipeline Result:")
print(nltk_pipeline(raw_text))
```

### spaCy Pipeline (Lemmatization)

```python
def spacy_pipeline(text):
    doc = nlp(text)

    lemmas = [
        token.lemma_.lower()
        for token in doc
        if token.is_alpha and not token.is_stop
    ]

    return lemmas

print("spaCy Pipeline Result:")
print(spacy_pipeline(raw_text))
```

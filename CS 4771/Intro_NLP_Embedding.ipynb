# 1. From Sparse Vector to Dense Representation

"Dense Representation," often called an "Embedding," was created precisely to solve the fundamental flaws of sparse representations like TF-IDF.

## 1.1 Why Use Dense Pepresentations?

The greatest and most fatal weakness of TF-IDF (and other sparse methods like One-Hot) is that they **cannot express the "similarity" relationship between words**.

In the world of TF-IDF (a space with tens of thousands of dimensions):

* The vector for `"king"`
* The vector for `"queen"`
* The vector for `"banana"`

The "distance" between any of these three vectors is almost **identical** (mathematically, they are "orthogonal"). The model cannot tell from the vectors that `"king"` is closer to `"queen"` than `"king"` is to `"banana"`.

This is the **"Semantic Gap."**

Dense Representations solve this problem in the following ways:

1.  **Capture Semantic Information**
    Dense representations (like Word2Vec) map words into a **low-dimensional** (e.g., 300 dimensions, not 50,000) **continuous vector space**. In this space, the **"direction" and "distance" of vectors are meaningful**.
    * Semantically similar words (like "king" and "queen") will have vectors that are very close in this space.
    * It can even learn complex relationships, such as the famous example:
        `vector("king") - vector("man") + vector("woman") ≈ vector("queen")`

2.  **Powerful Generalization**
    Because the model understands that "king" and "queen" are similar, if it has only seen "The king is rich" in the training data, when it encounters "The queen..." in a test, it can **automatically generalize** and guess that the "queen" might also be "rich." A TF-IDF model is completely incapable of this.

3.  **Efficiency and the Curse of Dimensionality**
    * **TF-IDF:** Very high-dimensional (equal to vocabulary size, e.g., 50,000) and mostly zeros (sparse). This is computationally wasteful and highly susceptible to the "Curse of Dimensionality."
    * **Dense:** Low-dimensional (e.g., 100, 300, 768) and nearly all values are non-zero floating-point numbers (dense). This makes it a **perfect input** for models like neural networks, and it is far more computationally efficient.

## 1.2 The History

### 1.2.1 Phase 1: Sparse Representation

* **Representative Tech:** One-Hot Encoding, Bag-of-Words (BoW), TF-IDF
* **Core Idea:** Based on "counts" and "frequency."
* **Characteristics:** High-dimensional, sparse, words are mutually independent (orthogonal).
* **Biggest Problem:** The Semantic Gap.

### 1.2.1 Phase 2: Matrix Factorization (Quasi-Dense Representation)

* **Representative Tech:** LSA (Latent Semantic Analysis)
* **Core Idea:** We are using a TF-IDF matrix, but this matrix is too sparse. We can use "matrix factorization" (like SVD) to **reduce its dimensionality**.
* **Evolution:** It attempts to discover "latent topics" from the "word-document" matrix. It does produce a denser vector and can find some synonyms (e.g., "car" and "auto" often appear in documents about the same topic), but the vectors lack clear physical meaning and are computationally expensive.

### 1.2.1 Phase 3: Static Dense Representations (Static Embeddings) - The Revolution Begins

* **Core Idea (The Critical Leap):** "You shall know a word by the company it keeps." — This is known as the **Distributional Hypothesis**.
* **Representative Tech:** **Word2Vec** (Google, 2013), **GloVe** (Stanford, 2014)
* **Evolution:**
    * **Word2Vec** (e.g., the Skip-gram model) stopped trying to use global statistics. Instead, it trains a small neural network: **given a center word (like "fox"), "predict" its surrounding context words (like "quick," "brown," "jumps")**.
    * To accomplish this "prediction" task, the neural network is **"forced"** to learn a low-dimensional vector (an embedding) for each word.
    * **Result:** This byproduct—the word vector—astonishingly captured semantics! This is where the `King - Man + Woman = Queen` example comes from.
* **Limitation (Static):** The word vectors from this phase are **"static."** The word "bank" has **only one** vector, regardless of the sentence. This cannot solve the **polysemy** (multiple meanings) problem.

### 1.2.1 Phase 4: Contextual Dense Representations (Contextual Embeddings) - The Current Era

* **Core Idea:** The meaning of a word **depends on its context**. The vector for "bank" in "river bank" and "money bank" should be **different**.
* **Representative Tech:** **ELMo** (2018), **BERT** (Google, 2018), **GPT** (OpenAI)
* **Evolution:**
    * **ELMo** used deep bidirectional LSTMs. A word's "vector" was no longer a fixed lookup, but the **internal state** of the network at that word's position after reading the whole sentence.
    * **BERT (Transformer)** completely abandoned LSTMs and used the **Attention Mechanism** to capture context even more powerfully.
* **Result:** We entered the era of "dynamic representations." A word's representation is dynamically generated based on the entire sentence it is in, perfectly solving the polysemy problem and dramatically improving performance on nearly all NLP tasks.

# 2. Word2Vec
In 2013, a team at Google led by Tomas Mikolov introduced **Word2Vec**. It's not a single algorithm, but a pair of related models: **CBOW** and **Skip-gram**.

Word2Vec is a *predictive* model. It uses a simple, shallow neural network to learn **dense** vector representations (called "embeddings") of words. Instead of a 50,000-dimensional sparse vector, a word might be represented by a 300-dimensional dense vector (e.g., `[0.23, -0.11, 0.45, ..., 0.89]`).

These dense vectors encode the semantic meaning learned from the contexts.

## 2.1 CBOW (Continuous Bag-of-Words)

The CBOW model's goal is to **predict a target word from its context words**.

* **Example Sentence:** "The quick brown fox jumps over the lazy dog."
* **Context Window (size 2):** `[The, quick, __, fox, jumps]`
* **Task:** Given the context words `['The', 'quick', 'fox', 'jumps']`, predict the center word `['brown']`.

## 2.2 Skip-gram

The Skip-gram model does the **exact opposite**: it **predicts the context words from a target word**.

* **Example Sentence:** "The quick brown fox jumps over the lazy dog."
* **Context Window (size 2):** `[The, quick, __, fox, jumps]`
* **Task:** Given the center word `['brown']`, predict the context words `['The', 'quick', 'fox', 'jumps']`.

**Which to use?**
* **Skip-gram:** Generally works better for large datasets and is particularly good at learning representations for rare words.
* **CBOW:** Is faster to train and works well for more frequent words.

# 3. GloVe (Global Vectors)
**GloVe (Global Vectors for Word Representation)** is an algorithm for learning word embeddings, developed by Stanford University in 2014.



## 3.1 The Core Idea

Its main goal was to combine the strengths of the two main families of embedding methods that existed at the time:

1.  **Global Matrix Factorization (e.g., LSA):** These methods were good at capturing **global statistics** of the entire corpus but performed poorly on "word analogy" tasks (e.g., $king - man + woman = queen$).
2.  **Local Context Window (e.g., Word2Vec):** These methods were excellent at "word analogy" tasks but **only saw local context** (a small window of words). They failed to leverage global statistics about how often words co-occur across the *entire* corpus.

GloVe's insight was that **you should train *directly* on the global statistics** in a way that captures the "predictive" power of Word2Vec.


GloVe revealed that the **ratio** of co-occurrence probabilities is a powerful source of meaning.

Consider the ratio $P(k | \text{ice}) / P(k | \text{steam})$ for different "probe" words $k$:

* If $k = \text{"solid"}$, this ratio will be **very large** (ice is solid, steam is not).
* If $k = \text{"gas"}$, this ratio will be **very small** (steam is gas, ice is not).
* If $k = \text{"water"}$ (related to both), the ratio will be **close to 1**.

GloVe is designed to produce word vectors whose relationships can *directly* encode these ratios.


## 3.2 The 2-Step Process
Therefore, the GloVe algorithm comprises two steps.


### 3.2.1 Build the Global Co-occurrence Matrix

This is the most important difference from Word2Vec.

1.  First, the algorithm scans the **entire corpus** one time.
2.  It builds a giant matrix **$X$**, where **$X_{ij}$** is the count of how many times word $j$ appeared in the context (e.g., a window of 10) of word $i$.
3.  *(Detail: GloVe uses a "distance-weighted" count, so words that are closer (distance 1) contribute more (1.0) to the count than words that are far (distance 4, contribution 1/4)).*

This matrix $X$ is the **"training data"** for GloVe.

### 3.2.2 Learn Vectors by "Fitting" the Matrix

GloVe does *not* use a predictive task like Skip-gram (i.e., "guess the context word").

Instead, it defines a **loss function** that says: "We must learn word vectors such that their dot product is related to their co-occurrence count."

The famous GloVe objective function is:

**$w_i \cdot \tilde{w}_j + b_i + \tilde{b}_j \approx \log(X_{ij})$**

Let's break this down:

* **$X_{ij}$**: The global co-occurrence count we built in Step 1.
* **$\log(X_{ij})$**: We use the log of the count, as counts can be massive.
* **$w_i$**: The **word vector** for the center word $i$ (This is what we want to learn!).
* **$\tilde{w}_j$**: The **context vector** for the context word $j$ (Just like in Word2Vec, each word has two temporary vectors).
* **$b_i, \tilde{b}_j$**: Two bias terms for stability.

The training process is a **weighted least-squares regression**. The model's only job is to adjust the values in $w$ and $\tilde{w}$ until this "approximation" ($\approx$) holds true for all the non-zero entries in the matrix $X$.

The **weighting function** $f(X_{ij})$ in the full loss function is also crucial: it gives *less* weight to extremely common word pairs (like "the" and "is") and *zero* weight to pairs that never co-occur. This is very similar to how Word2Vec's negative sampling avoids high-frequency words.


### 3.2.3 The Vectors

After training is complete, we have two sets of vectors: the $W$ matrix (center words) and the $\tilde{W}$ matrix (context words).

The GloVe authors found that the **best final embedding** for each word $i$ is:

**$W_{\text{final}} = W + \tilde{W}$**

```python
# This will download the GloVe model trained on Wikipedia (about 160MB)
# Common models: 'glove-wiki-gigaword-100', 'word2vec-google-news-300'
try:
    glove_model = gensim.downloader.load('glove-wiki-gigaword-100')
    print("Gensim GloVe model loaded successfully.")
except Exception as e:
    print(f"Error loading model: {e}")
    print("Please check your internet connection.")
```

```python
# Find the most similar words
print("--- Most similar to 'king' ---")
try:
    display(glove_model.most_similar('king'))
except NameError: # In case 'glove_model' failed to load
    print("Model not loaded, skipping.")
```

```python
# Perform the famous vector arithmetic: King - Man + Woman = ?
print("\n--- 'king' - 'man' + 'woman' ---")
try:
    display(glove_model.most_similar(positive=['king', 'woman'], negative=['man'], topn=5))
except NameError:
    print("Model not loaded, skipping.")
```

```python
# Find the word that doesn't match
print("\n--- Word that doesn't match ---")
try:
    display(glove_model.doesnt_match(['breakfast', 'cereal', 'dinner', 'computer']))
except NameError:
    print("Model not loaded, skipping.")
```

## 4.2 Train a Simple Word2Vec Model with `gensim`

Sometimes, pre-trained models aren't enough. If you have a very specific domain (e.g., medical texts, legal documents, project-specific jargon), the general-purpose embeddings won't know your specific vocabulary. In this case, you can train your own model.

```python
from gensim.models import Word2Vec
```

```python
# Define your corpus. It must be a list of lists of tokens (words).
corpus = [
    ['the', 'quick', 'brown', 'fox', 'jumps', 'over', 'the', 'lazy', 'dog'],
    ['a', 'fast', 'brown', 'fox', 'is', 'not', 'a', 'lazy', 'dog'],
    ['the', 'king', 'and', 'queen', 'sat', 'on', 'the', 'throne'],
    ['the', 'prince', 'is', 'the', 'son', 'of', 'the', 'king'],
    ['the', 'princess', 'is', 'the', 'daughter', 'of', 'the', 'queen'],
    ['a', 'man', 'and', 'a', 'woman', 'walked', 'into', 'the', 'palace'],
    ['the', 'prince', 'will', 'become', 'a', 'king'],
    ['the', 'princess', 'will', 'become', 'a', 'queen']
]

print(f"Corpus size: {len(corpus)} sentences")
```

```python
# 2. Train the model
model = Word2Vec(
    sentences=corpus,      # Our corpus
    vector_size=50,      # Dimensionality of the embedding (low for our small corpus)
    window=3,            # Context window size
    min_count=1,         # Ignore words with frequency lower than this
    workers=4,           # Number of CPU cores to use
    sg=1                 # Use Skip-gram (1) or CBOW (0). Let's use Skip-gram.
)

print("Model trained successfully!")
```

```python
# Get the vocabulary
words = list(model.wv.key_to_index.keys())
print(f"Vocabulary ({len(words)} words):\n{words}\n")

# Check most similar words
print("--- Most similar to 'king' (custom model) ---")
try:
    display(model.wv.most_similar('king', topn=5))
except Exception as e:
    print(f"Error: {e}. 'king' might not be in vocab or model is too small.")
```

```python
# Try the vector arithmetic
print("\n--- 'king' - 'man' + 'woman' (custom model) ---")
try:
    display(model.wv.most_similar(positive=['king', 'woman'], negative=['man'], topn=5))
except Exception as e:
    print(f"Could not perform arithmetic: {e}. Needs all words in vocab.")

# Let's see what it learned for 'fox'
print("\n--- Most similar to 'fox' (custom model) ---")
try:
    display(model.wv.most_similar('fox', topn=5))
except Exception as e:
    print(f"Error: {e}. 'fox' might not be in vocab or model is too small.")
```
